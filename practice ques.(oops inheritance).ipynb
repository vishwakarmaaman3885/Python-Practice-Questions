#### Practice questions on inheritance

```python
# Create a Person class with a name and display_name() method. Create a Student class that inherits from Person and adds course.
class Person:
    def __init__(self,name):
        self.name = name
    def display_name(self):
        print("my name is",self.name)
class Student(Person):
    def __init__(self, name,course):
        super().__init__(name)
        self.course = course

student = Student("Aman","DS")
student.display_name()
print("Course",student.course) 
```

```python
# Create a Vehicle class with brand and show_brand(). Create a Car class that inherits from Vehicle and adds model.
class Vehicle:
    def __init__(self,brand):
        self.brand = brand
    def show_brand(self):
        print("my brand is",self.brand)
class Car(Vehicle):
    def __init__(self, brand,model):
        super().__init__(brand)
        self.model = model
car = Car("TATA","Nexon")
car.show_brand()
print("Model",car.model)        
```

```python
# Multilevel Inheritance
# Create:
# Animal → Dog → Puppy
# Animal → eat()
# Dog → bark()
# Puppy → play()
class Animal:
    def eat(self):
        print("Eating")
class Dog(Animal):
    def bark(self):
        print("Barking")
class Puppy(Dog):
    def play(self):
        print("Playing")
p1 = Puppy()
p1.play() 
p1.bark()
p1.eat() 
```

```python
# Multiple Inheritance
# Father → skills()
# Mother → hobby()
# Child(Father, Mother)
class Father:
    def skills(self):
        print("Driving")
class Mother:
    def hobby(self):
        print("Painting")
class child(Father,Mother):
    pass
c1 = child()
c1.hobby() 
c1.skills() 
```

```python
# Create Dog that inherits from Animal and overrides sound() to print "Dog barks".
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
d1 = Dog()
d1.sound() 
```
