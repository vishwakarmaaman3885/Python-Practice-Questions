```python
# 2.Use a for loop to calculate the average of the numbers.
numbers = [10, 15, 20, 25, 30, 35, 40]
total = 0
count = 0
for i in numbers:
    total = total+i
    count = count+1
average = total/count
print(average) 
```

```python
# Using a for loop, count how many numbers are greater than 25.
numbers = [10, 15, 20, 25, 30, 35, 40]
count = 0
for i in numbers:
    if i >25:
        count = count+1
print(count) 
```

```python
# Calculate the sum of only the numbers greater than 25.
numbers = [10, 15, 20, 25, 30, 35, 40]
count = 0
total = 0
for i in numbers:
    if i>25:
        count = count+1
        total = total+i
print(total) 
```

```python
# Using a for loop, calculate the average of only the numbers greater than 25.
numbers = [10, 15, 20, 25, 30, 35, 40]
count = 0
total = 0
for i in numbers:
    if i>25:
        count = count+1
        total = total+i
average = total/count
print(average) 
```

```python
# Using a for loop, find the maximum number without using:
numbers = [10, 15, 20, 25, 30, 35, 40]
# print(max(numbers))
# print(sorted(numbers)) 
max = numbers[0]
for i in numbers:
    if i >max:
        max = i
print(max) 
```

```python
# Find the minimum number without using min(), sort(), or sorted():
numbers = [45, 12, 78, 23, 91, 34, 67]
min = numbers[0]
for i in numbers:
    if i<min:
        min = i
print(min) 
```

```python
# Find the second largest number without using:
numbers = [45, 12, 78, 23, 91, 34, 67]
max = numbers[0]
second_max = numbers[0]
for i in numbers:
    if i>max or i<second_max:
        max = i
        second_max=i
print(max) 
```

```python
numbers = [45, 12, 78, 23, 91, 34, 67]
largest = numbers[0]
second_largest = numbers[0]
for i in numbers:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest:
        second_largest = i
print(second_largest)
```

```python
# Count how many times each number appears in a list.
numbers = [2, 3, 2, 5, 3, 2, 7, 5, 3]
frequency = {}
for i in numbers:
    if i in frequency:
        frequency[i] = frequency[i]+1
    else:
        frequency[i] = 1
print(frequency)  
```
