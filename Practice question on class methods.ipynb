#### classmethod practice questions

```python
# Create a Student class with a class variable:
# Create a class method change_school() that changes the school name.
class student:
    school = "ABC"
    def __init__(self,name):
        self.name=name
    @classmethod
    def change_school(cls,new_school):
        student.school = new_school
    def school_details(self):
        print(self.name,self.school)  

s1 = student("Aman")
s1.school_details()
s1.change_school("XYZ")
s1.school_details() 
```

```python
# Create a BankAccount class with:
# Create a class method change_bank() that changes the bank name.
# bank_name = "SBI"
class bankaccount:
    bank_name = "SBI"
    def __init__(self,name):
        self.name = name
    @classmethod
    def change_bank(cls,new_bank):
        cls.bank_name = new_bank
    def bank_details(self):
        print(self.name,self.bank_name)

b1 = bankaccount("Aman")
b2 = bankaccount("Rahul")
bankaccount.change_bank("HDFC")
b1.bank_details()
b2.bank_details() 
```

```python
# Create a Student class:
# Create a class method called from_string() that can create a Student object from:
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    @classmethod
    def from_string(cls,data):
        name,age = data.split("-")
        return cls(name,int(age))
s1 = student.from_string("Aman-20")
print(s1.name)
print(s1.age) 
```
