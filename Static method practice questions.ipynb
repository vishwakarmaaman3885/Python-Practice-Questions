### Practice questions on static method

```python
# Create an Employee class with a static method:
# The method should return True if the email:
# contains @
# contains . after @
# Otherwise return False.
class employee:
    @staticmethod
    def validate_email(email):
        if "@" not in email or "." not in email:
            return False
        
        at_position = email.index("@")
        dot_position = email.index(".")
        
        if dot_position>at_position:
            return True
        return False
print(employee.validate_email("aman@gmail.com"))
print(employee.validate_email("aman@gmail")) 
print(employee.validate_email("aman.com@gmail")) 
```

```python
# Create a Product class with a static method:
# The method should:
# Calculate the discount amount.
# Return the final price after discount.
class product:
    @staticmethod
    def actual_price(price):
        if price >= 5000:
            discount = price*0.3
        else:
            discount = price*0.1
        
        final_price = price-discount
        return final_price

print(product.actual_price(6000))
print(product.actual_price(3000))       
```

```python
# Create an Employee class:
# Salary must be greater than 0
# Salary must not exceed 10,00,000
# Return True if valid, otherwise False.
class employee:
    @staticmethod
    def actual_salary(salary):
        if salary>0 and salary<=1000000:
            return True
        else:
            return False
print(employee.actual_salary(12000)) 
print(employee.actual_salary(-12000))
```
