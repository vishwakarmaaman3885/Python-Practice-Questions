### Some practice questions on functions

```python
# Q1. Write a function greet(name) that prints:
def greet(name):
    return "Hello "+ name
print(greet('Aman'))  
```

```python
# Q2. Write a function is_even(n) that returns True if a number is even and False otherwise.
def is_even(a):
        if a%2==0:
            return True
        else:
            return False
print(is_even(1)) 
```

```python
# Q3. Write a function find_largest(numbers) to find the largest number without using max().
# find_largest([10, 25, 8, 45, 30])
def find_largest(a):
    largest = a[0] # Finding largest/smallest: than a[0]
    for i in a:
        if i>largest:
            largest = i
    return largest
print(find_largest([10,25,8,45,30]))  
```

```python
# Q4. Write a function count_vowels(text) that counts a, e, i, o, u.
# count_vowels("Data Science")
def count_vowels(a):
    count = 0
    vowels = ["a","e","i","o","u"]
    for i in a:
        if i in vowels:
            count = count+1
    return count
print(count_vowels("Data Science"))  
```

```python
# Q5. What is the output?
def test(*args):
    print(args)
    print(type(args))
test(10, 20, 30) 
```

```python
# Q6. Write a function calculate_sum(*args) that returns the sum of any number of numbers.
# calculate_sum(10, 20, 30, 40)
def calculate_sum(*args):
    total = 0
    for i in args:
        total = total+i
    return total
print(calculate_sum(10, 20, 30, 40)) 
```

```python
# Q7. Write a function second_largest(*args) that finds the second-largest number without using sort() or max().
# second_largest(10, 45, 20, 80, 30)
def second_largest(*args):
    largest = args[0]
    second_largest = args[0]
    for i in args:
        if i>largest:
           second_largest=largest
           largest = i
        elif i>second_largest:
            second_largest=i
    return second_largest
print(second_largest(10, 45, 20, 80, 30))  
```

```python
# **kwargs
# Q8. What is the output?
def student(**kwargs):
    print(kwargs)
    print(type(kwargs))
student(name="Aman", age=20, course="Data Science") 
```

```python
# Q9. Write a function student_info(**kwargs) that prints every key and value.
# student_info(name="Aman", age=20, city="Indore")
def student_info(**kwargs):
    for key,value in kwargs.items():
        print(key,value)
student_info(name="Aman", age=20, city="Indore") 
```

```python
# Q10. Write a function get_marks(**kwargs) that returns the total marks.
# get_marks(Python=80, SQL=90, Maths=70)
def get_marks(**kwargs): 
    total = 0
    for i in kwargs.values():
        total = total+i
    return total
print(get_marks(Python=80, SQL=90, Maths=70)) 
```

```python
# Code	Gives you
# kwargs	keys
# kwargs.keys()	keys
# kwargs.values()	values
# kwargs.items()	key + value
```

```python
# Q11. What is the output
def test(*args, **kwargs):
    print(args)
    print(kwargs)
test(10, 20, 30, name="Aman", age=20)
```

```python
# Q12. Create a function that accepts numbers through *args and a student's name through **kwargs, then prints the average marks and name.
# student(80, 90, 70, name="Aman")
def student(*args,**kwargs):
    total = 0
    count = 0
    for i in args:
        total = total +i
        count = count+1
    average = total/count
    print("Name:", kwargs["name"])
    print("Average:", average)

student(80, 90, 70, name="Aman")
```

```python
# yield / Generators
# Q13. What is the output?
def numbers():
    yield 10
    yield 20
    yield 30
x = numbers()
print(next(x))
print(next(x)) 
```

```python
# Q14. Write a generator even_numbers(n) that yields even numbers from 1 to 10
# for i in even_numbers(10):
#     print(i)
def even_numbers(n):
    for i in range(1,n+1):
        if i %2==0:
            yield i
for i in even_numbers(10):
    print(i) 
```

```python
# Create a generator that accepts any number of values using *args and yields only integers.
# data = generate(10, "Aman", 20, 4.5, 30, "Python", 40)
# for i in data:
#     print(i)
def generate(*args):
    for i in args:
        if isinstance(i,int):
            yield i
data = generate(10, "Aman", 20, 4.5, 30, "Python", 40)
for i in data:
    print(i) 
```
