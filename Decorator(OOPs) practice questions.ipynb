#### Decorator(OOPs) - practice questions 

```python
# Create a decorator that prints "Function started" before calling the function and "Function ended" after it.
def my_deco(func):
    def inner_deco():
        print("function started")
        func()
        print("function ended")
    return inner_deco
```

```python
@ my_deco
def test1():
    print("Hello") 
```

```python
test1() 
```

```python
# Create a decorator that works with a function having arguments:
def deco(func):
    def inner_deco(*args): # *args = "Take any number of positional arguments."
        print('Function started')
        func(*args)
        print("Function ended")
    return inner_deco 
```

```python
@deco
def add(a, b):
    print(a + b) 
```

```python
add(34,56) 
```

```python
@deco
def add(a, b,c):
    print(a + b+c)  
```

```python
add(34,56,45)  
```
