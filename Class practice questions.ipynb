```python
# Q4. Create a Rectangle class with length and width. Create a method to calculate and return its area.
class rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def returns_area(self):
        return self.length * self.width

values = rectangle(12,10)
values.returns_area() 
```

```python
# Q5. Create a BankAccount class with balance. Create methods:

# deposit()
# withdraw()
# display_balance()
class Bankaccount:
    def __init__(self,balance):
        self.balance = balance
    
    def deposit(self,amount):
        self.balance = self.balance+amount
        return self.balance
    
    def withdraw(self,amount):
        self.balance = self.balance-amount
        return self.balance
    
    def display_balance(self):
        return self.balance
    
       
account = Bankaccount(1000)
account.deposit(500)
account.withdraw(300)

account.display_balance() 
```

```python
# Q6. Create a Calculator class with methods for:

# addition
# subtraction
# multiplication
# division

class calculator:
    def add(self,a,b):
        return a+b
    def subtract(self,a,b):
        return a-b
    def multiply(self,a,b):
        return a*b
    def divison(self,a,b):
        return a/b
        
calc = calculator()

print(calc.add(10,17)) 
print(calc.subtract(23,4))
print(calc.multiply(23,4))
print(calc.divison(20,4))     
```

```python
# Q7. Create a Student class that stores name and marks in 3 subjects. Create a method that calculates and returns
# the total and average marks.
class marks:
    def __init__(self,name,maths_marks,python_marks,sql_marks):
        self.name = name
        self.maths_marks = maths_marks
        self.python_marks = python_marks
        self.sql_marks = sql_marks
    def calculate(self):
        total = self.maths_marks + self.python_marks+ self.sql_marks
        average = total/3
        return self.name, total,average
    
student = marks("Aman",67,78,90)
student.calculate() 
```

```python
# Q8. Create an Employee class with name and salary. Create a method that increases the salary by 10%.
class employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def calculate(self):
        salary = self.salary+self.salary*0.10
        return self.name, salary

student = employee("Aman",50000)
student.calculate() 
```

```python
# Q9. Create a Person class with a method display(). Create a Student class that inherits from Person and adds a course attribute.
class product:
    def __init__(self,name,price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    def return_total_price(self):
        return self.price*self.quantity

product = product("Aman",50000,2)
product.return_total_price() 
```

```python
# Q10. Create a Vehicle class with a start() method. Create Car and Bike classes that inherit from Vehicle and override the start() method.
class vehicle:
    def start(self):
        return "Vehicle is starting"
    
class car(vehicle):
    def start(self):
        return "car is starting"
    
class bike(vehicle):
    def start(self):
        return "bike is starting"
    
car = car()
bike = bike()
print(car.start())
print(bike.start()) 
```
