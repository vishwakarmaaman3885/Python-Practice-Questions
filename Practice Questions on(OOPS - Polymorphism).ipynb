#### OOPS Practice on polymorphism

```python
#Q1. Same method, different classes
# Create two classes Dog and Cat.
class Dog:
    def sound(self):
        return "Dog says Woof"

class Cat:
    def sound(self):
        return "Cat says Meow"
    
dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound()) 
```

```python
# Q2. Same method with different behavior
# Car.start() should return "Car starts" and Bike.start() should return "Bike starts".

class Car:
    def sound(self):
        return "Car Starts"

class Bike:
    def sound(self):
        return "Bike Starts"

car = Car()
bike = Bike()

print(car.sound())
print(bike.sound()) 
```

```python
# Q3. Polymorphism using a loop
# Create Dog and Cat classes, both having sound().
# Use one for loop to call sound() for both objects.
class Dog:
    def sound(self):
        return "Dog says Woof"

class Cat:
    def sound(self):
        return "Cat says Meow"

dog = Dog()
cat = Cat()

animal = [dog,cat]

for animal in animal:
    print(animal.sound()) 
```

```python
# Q4. Different classes, same method
# Create objects and call details().

class Student:
    def details(self):
        return "Student details"
    
class Teacher:
    def details(self):
        return "Teacher details"
    
student = Student()
teacher = Teacher()

print(student.details())
print(teacher.details()) 
```

```python
# Q5. Function + different objects
# The function should call shape.area().
# Use it for both Circle and Rectangle.
class Circle:
    def area(self):
        return 3.14*5*5
    
class Rectangle:
    def area(self):
        return 10*5
    
def calculate_area(shape):
    return shape.area()

circle = Circle()
rectangle = Rectangle()

print(calculate_area(circle))
print(calculate_area(rectangle)) 
```

```python
# Q6. Polymorphism with a list
# Then use a loop to call syllabus().
class DataScience:
    def syllabus(self):
        return "Data Science syllabus"

class Webdevelopment:
    def syllabus(self):
        return "Web Development syllabus"
    
class Machinelearning:
    def syllabus(self):
        return "Machine learning syllabus"
    
ds = DataScience()
web = Webdevelopment()
ml = Machinelearning()

courses = [ds,web,ml]

for course in courses:
    print(course.syllabus()) 
```

```python
# Q7. Duck typing
# The function should work with all three objects without checking their class using if.
class Excel:
    def open(self):
        return "Excel"
class PowerBI:
    def open(self):
        return "Power BI"
class Python:
    def open(self):
        return "Python"
exl = Excel()
pbi = PowerBI()
py = Python()

subjects = [exl,pbi,py]
for subject in subjects:
    print(subject.open())  
```

```python
# Q8. Polymorphism with parameters
# Both should have the same method name but produce different messages.
class Dog:
    def eat(self,food):
        return f"Dog eats {food}"
class Cat:
    def eat(self,food):
        return f"Cat eats {food}"

dog = Dog()
cat = Cat()
animals = [dog,cat] 
for animal in animals:
    print(animal.eat("meat")) 
```

```python
# Q9. Inheritance + method overriding
# Create a parent class:
class Employee:
    def work(self):
        return "Employee is working"
class Devloper(Employee):
    def work(self):
        return "Developer is writting code"
class DataScientist(Employee):
    def work(self):
        return "Data Scientist creating a model"

devloper = Devloper()
dataScientist = DataScientist()

employees = [devloper,dataScientist]

for employee in employees:
    print(employee.work()) 
```

```python
# Q12. Important interview question 
# What is the difference between:
# Method Overriding and Polymorphism?
# Give one example of each.

# Overriding = what the child class does.
# Polymorphism = how the same method call can behave differently for different objects.

#1. Method Overriding

class Employee:
    def work(self):
        return "Employee is working"
class Devloper(Employee):
    def work(self):
        return "Developer is writing code"

devloper = Devloper()
print(devloper.work()) 
```

```python
# 2. Polymorphism
class Developer:
    def work(self):
        return "Developer is writting code"
    
class DataScientist:
    def work(self):
        return "Data scientist is creating model"

employees = [Devloper(),DataScientist()]

for employee in employees:
    print(employee.work()) 
```

```python
# Q13. Real-world interview problem
# Create a list containing all three objects.
class DataScientist:
    def work(self):
        return "Data Science"

class MLEngineer:
    def work(self):
        return "ML Engineer"

class DataAnalyst:
    def work(self):
        return "Data Analyst"

subjects = [DataScientist(),MLEngineer(),DataAnalyst()]

def employee_work(subjects):
    for subject in subjects:
        print(subject.work())

employee_work(subjects) 
```

```python
# Q13. Real-world interview problem 
# Here is a real-world polymorphism problem. Try it yourself first.
# Problem: Payment System 
# Create three classes:
# CreditCard
# UPI
# PayPal

class CreditCard:
    def pay(self, amount):
        return "Paid ₹" + str(amount) + " using Credit Card"

class UPI:
    def pay(self, amount):
        return "Paid ₹" + str(amount) + " using UPI"

class PayPal:
    def pay(self, amount):
        return "Paid ₹" + str(amount) + " using PayPal"

payments = [
    CreditCard(),
    UPI(),
    PayPal()
]
def process_payment(payments):
    for payment in payments:
        print(payment.pay(500))

process_payment(payments)
```
