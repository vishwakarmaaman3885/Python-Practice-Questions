#### lambda,map,reduce,filter practice questions

#### lambda ques.

```python
# Q1. Create a lambda function to find the square of a number.
# square(5)
square = (lambda x:x**2)
square(5) 
```

```python
# Create a lambda function to find the maximum of two numbers.
# maximum(10, 20)
maximum = (lambda x,y:x if x > y else y)
maximum(10,20) 
```

```python
# Create a lambda function to check whether a number is even or odd.
# check(10)
even_odd = (lambda x: "even" if x %2==0 else "odd")
even_odd(10)  
```

```python
# Create a lambda function to find the largest of three numbers.
# largest(10, 25, 15)
largest = (lambda x,y,z:x if x>y and x>z else y if y>x and y>z else z)
largest(10,25,15) 
```

#### map-transform the value

```python
# Using map(), create a list containing the squares of:
numbers = [1, 2, 3, 4, 5]
list(map(lambda x: x**2 ,numbers)) 
```

```python
# Using map(), convert:
names = ["aman", "raj", "rohit"]
list(map(lambda x:x.upper(), names)) 
```

```python
# Using map(), add corresponding elements.
l1 = [1, 2, 3, 4, 5]
l2 = [6, 7, 8, 9, 10]
list(map(lambda x,y:x+y,l1,l2)) 
```

```python
# Using map(), multiply corresponding elements of l1 and l2.
# [6, 14, 24, 36, 50]
list(map(lambda x,y: x*y,l1,l2))
```

#### filter-select the data

```python
# Using filter(), extract all even numbers:
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
list(filter(lambda x: x%2==0,numbers)) 
```

```python
# Using filter(), extract numbers greater than 50:
numbers = [20, 55, 30, 80, 45, 90]
list(filter(lambda x: x>50 , numbers)) 
```

```python
# Using filter(), extract strings whose length is greater than 4:
names = ["Aman", "Rahul", "Raj", "Rohit", "Vishal"]
list(filter(lambda x: len(x)>4,names ))  
```

#### reduce()-combine the (single list)

```python
from functools import reduce
```

```python
# Using reduce(), calculate the sum:
numbers = [1, 2, 3, 4, 5]
reduce(lambda x,y:x+y,numbers) 
```

```python
# Using reduce(), calculate the product:
numbers = [1, 2, 3, 4, 5]
reduce(lambda x,y:x*y,numbers) 
```

```python
# Using reduce(), find the largest number without using max():
numbers = [10, 45, 20, 80, 30]
reduce(lambda x,y: x if x>y else y, numbers) 
```

```python
# Filter only even numbers.
# Square those even numbers.
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_num = list(filter(lambda x: x%2==0,numbers))
square = list(map(lambda x:x**2,even_num))
print(square) 
```

```python
# Use filter() + map() + reduce() together.
# 1. Select even numbers
# 2. Square them
# 3. Calculate their total
numbers = [1, 2, 3, 4, 5, 6]
even_num = list(filter(lambda x:x%2 ==0,numbers))
square = list(map(lambda x:x**2,even_num))
total = reduce(lambda x,y:x+y,square)
print(total) 
```
