##### Encapsulation practice questions

```python
# Create a class Student.
# Private variable: __marks
# set_marks(marks)
# get_marks()
# Do not allow marks below 0 or above 100.

# set_...() → puts data inside the object
# get_...() → takes data out of the object

class student:
    def __init__(self):
        self.__marks = 0  # __marks is private
    def set_marks(self,marks):
        if marks>=0 and marks <= 100: 
            self.__marks = marks
        else:
            print("Marks must between 0 to 100")
    def get_marks(self):
        return self.__marks
s1 = student()
s1.set_marks(80) # A setter is used to store or update a value.
print(s1.get_marks())  # A getter is used to read/retrieve a value.
```

```python
# 2. Bank Account-Create a class BankAccount.
# Private variable: __balance, deposit(amount), withdraw(amount), get_balance(), negative value is not allow
class bank_account:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self,amount):
        self.__balance = self.__balance + amount
    def withdraw(self,amount):
        if self.__balance >= amount:
            self.__balance = self.__balance-amount
            return "Transaction is successful"
        else:
            return "Insufficient balance"
    def get_balance(self):
        return self.__balance 
```

```python
obj_bank_account = bank_account(1000)
```

```python
obj_bank_account.get_balance() 
```

```python
obj_bank_account.withdraw(500) 
```

```python
obj_bank_account.get_balance() 
```

```python
# 5. Mobile Phone Create a class Mobile.Private variable: __battery,charge(percent),use(percent),get_battery()
# Battery should always stay between 0 and 100.
class mobile:
    def __init__(self):
        self.__battery = 0
    def charge(self,percent):
        if self.__battery + percent <=100:
            self.__battery = self.__battery + percent
        else:
            return "Battery can't exceed 100"
    def use(self,percent):
        if self.__battery  - percent >= 0:
            self.__battery = self.__battery  - percent
        else:
            return "Battery can't go below 0"
    def get_battery(self):
        return self.__battery
m1 = mobile() 
    
```

```python
m1.charge(80)
```

```python
m1.use(40) 
```

```python
m1.get_battery() 
```

```python
# ATM Machine-Create a class ATM.Private variable: __pin,change_pin(old_pin, new_pin),verify_pin(pin)
# Only change the PIN if the old PIN is correct.
class ATM:
    def __init__(self):
        self.__pin = 12345
    def change_pin(self,old_pin,new_pin):
        if self.__pin == old_pin:
            self.__pin = new_pin
            return "PIN changed successfully"
        else:
            return "Old PIN is incorrect"
    def verify_pin(self,pin):
        return self.__pin == pin
a1 = ATM() 
```

```python
a1.verify_pin(12345)
```

```python
a1.change_pin(12345,67890)
```

```python
a1.verify_pin(67890) 
```

```python
# Data Science Dataset-Create a class Dataset.Private variable: __rows,add_rows(n),remove_rows(n),get_rows(),Rows should never 
# become negative.
class Dataset:
    def __init__(self):
        self.__rows = 0
    def add_rows(self,n):
        if n >= 0:
            self.__rows = self.__rows+n
        else:
            print("Rows can't be negative") 
    def remove_rows(self,n):
        if n<= self.__rows:
            self.__rows = self.__rows - n
        else:
            print("Rows can't be negative")
    def get_rows(self):
        return self.__rows
d1 = Dataset()
d1.add_rows(5)
d1.remove_rows(2) 
d1.get_rows() 
```

```python
# E-commerce Cart (Interview Level),Create a class ShoppingCart.Private variable: __total, add_item(price),remove_item(price),get_total()
# Total should never become negative.
class ShoppingCart:
    def __init__(self):
        self.__total = 0
    def add_item(self,price):
        if price>=0:
            self.__total = self.__total + price
        else:
            print("total should never become negative") 
    def remove_item(self,price):
        if  self.__total >= price:
            self.__total = self.__total - price
        else:
            print("total should never become negative") 
    def get_total(self):
        return self.__total
cart = ShoppingCart()
cart.add_item(45)
cart.remove_item(21)
cart.get_total() 
```
