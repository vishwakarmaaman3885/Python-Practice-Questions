### Sets practice questions

```python
# A set is a collection of unique items.- Mutable

# Create a set of 5 colors and print it.
colors = {"Red","Blue","Green","White","Black"}
colors
```

```python
# Add "orange" to:
colors = {"red", "blue", "green"}
colors.add('orange')
print(colors) 
```

```python
# Remove "blue" from the set.
colors = {"red", "blue", "green"}
colors.remove('blue')
colors
```

```python
colors = {"red", "blue", "green"}
colors.discard('blue')
colors
```

```python
# Find the common elements:
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a&b) 
```

```python
# Find all unique elements:
a = {1, 2, 3}
b = {3, 4, 5}
print(a|b) 
print(set(a) | set(b)) # union- all unique value
```

```python
# Remove duplicates:
numbers = [1, 2, 2, 3, 4, 4, 5]
print(set(numbers)) 
```

```python
# To remove duplicates and keep the original order, use a loop:
numbers = [3, 1, 3, 2, 1, 4, 2]
unique = []
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique) 
```

```python
numbers = [5, 2, 5, 3, 2, 7, 3]
unique = set(numbers)
print(sorted(unique))
```

#### Tuple- A tuple is a collection of values that is ordered and immutable.

```python
# Create a tuple containing 5 fruits and print the third fruit.
fruits = ("apple","orange","grapes",'mango','kiwi')
fruits[2] 
```

```python
# Print the first and last element.
numbers = (10, 20, 30, 40, 50)
print(numbers[0])
print(numbers[-1]) 
```

```python
# Count how many times 2 appears:
nums = (1, 2, 3, 2, 4, 2)
count = 0
for i in nums:
    if i==2:
       count = count+1
print(count) 
```

```python
# count how many times each numbers appear
nums = (1, 2, 3, 2, 4, 2)
count = {}  # {} → empty dictionary
for i in nums:
    if i in count:
        count[i] = count[i]+1
    else:
        count[i] = 1
print(count) 
```

```python
# Find the index of "Python":
languages = ("Java", "Python", "C++", "SQL")
print(languages.index('Python'))  # tuple.index(value) → finds the position of a value.
```

#### List-A list is a collection used to store multiple values in a single variable.

```python
# Find the largest number without using max().
numbers = [45, 12, 78, 23, 91, 34, 67]
max = 0
for i in numbers:
    if i>max:
        max = i
print(max)  
```

```python
# Find the second largest number.
numbers = [45, 12, 78, 23, 91, 34, 67]
max = 0
second_max = 0 
for i in numbers:
    if i >max :
        second_max = max
        max = i 
    elif i>second_max:
        second_max = i
print(second_max)   
```

```python
numbers = [45, 12, 78, 23, 91, 34, 67]
sorted_numbers = sorted(numbers)
print(sorted_numbers[-2]) 
```

```python
# Remove duplicates from a list.
numbers = [1, 2, 2, 3, 4, 4, 5, 5]
unique = list(set(numbers)) 
unique
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5]
unique = []
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique) 
```

```python
# Count frequency of each number.
numbers = [1, 2, 2, 3, 1, 4, 2, 3]
frequency = {}
for i in numbers:
    if i in frequency:
        frequency[i] = frequency[i]+1
    else:
        frequency[i] = 1
print(frequency) 
```

```python
# Find even and odd numbers.
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
for i in numbers:
    if i%2==0:
        print(i,'even')
    else: 
        print(i,'odd')  
```

```python
# Find the sum of all elements without using sum().
numbers = [45, 12, 78, 23, 91, 34, 67]
total = 0
for i in numbers:
    total = total+i
print(total) 
```

```python
# Reverse a list without using .reverse().
numbers = [45, 12, 78, 23, 91, 34, 67]
print(numbers[::-1]) 
```

```python
numbers = [45, 12, 78, 23, 91, 34, 67]
numbers.reverse() 
print(numbers)
```

```python
# Find common elements between two lists.
a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]
common = []
for i in a:
    if i in b:
        common.append(i)
print(common) 
```

```python
# Find the missing number.
numbers = [1, 2, 3, 5, 6]
for i in range(1,7):
    if i not in numbers:
      print(i) 
        
```

```python
# Move all zeros to the end.
numbers = [0, 1, 0, 3, 12]
result = []
for i in numbers:
    if i!=0:
        result.append(i)
print(result) 
```
