#### Some pattern matching questions

```python
# Square
n=6
for i in range(n):
    for j in range(n):
      print("*",end="")
    print("\r") 
```

```python
# Right triangle
n=6
for i in range(0,n):
    for j in range(0,i+1):
      print("*",end="")
    print("\r") 
```

```python
# Inverted right triangle
n = 6
for i in range(n):
    for j in range(i):
        print("",end="")
    for j in range(n-i):
        print("*",end="")
    print() 
```

```python
# pyramid pattern
n=6
for i in range(1,n+1):
    for j in range(2*i-1):
        print("*",end="")
    print("\r") 
```

```python
# Inverted Pyramid
n=6
for i in range(n):
    for j in range(i):
        print("",end="")
    for j in range(2*(n-i)-1):
        print("*",end="")
    print("\r")
```

```python
# Diamond Pattern
# upper half
n = 5
for i in range(n):
    for j in range((n-i)-1):
        print("",end="")
    for j in range(2*i+1):
        print("*",end="") 
    print("\r") 
# lower half
for i in range(n-1):
    for j in range(i+1):
        print("",end="")
    for j in range(2*(n-i)-3):
        print("*",end="") 
    print() 
```

```python
# Hollow square
n=5
for i in range(n):
    for j in range(n): # i = 0,4 gives 5 star
        if i ==0 or i==n-1 or j==0 or j==n-1: # 0,1,2,3,4= n-1=4,j=0
            print("*",end="") #j:      0   1   2   3   4 ↓output: *   ␠   ␠   ␠   *
        else:
            print(" ",end="")  # j-check the value if value is statisfied than 5 star else espace
    print()
```

```python
# Hollow pyramid
n=5
for i in range(n):
    #spaces before stars
    for j in range((n-i)-1):
        print(" ",end="") 
    # stars and spaces
    for j in range(2*i+1):
        if j==0 or j==2*i or i ==n-1:
            print("*",end="")
        else:
            print(" ",end="")
    print()     
```

```python
n = 5
for i in range(n):
    for j in range(n - i - 1):
        print("", end="")

    for j in range(2 * i + 1):

        if j == 0 or j == 2 * i or i == n - 1:
            print("*", end="")
        else:
            print("", end="")

    print()
```

```python
# Butterfly
n = 4
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    for j in range(2 * (n - i)):
        print(" ", end="")
    for j in range(i):
        print("*", end="")
    print()
```

```python
# Number + star pattern
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```
