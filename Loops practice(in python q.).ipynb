### LOOP

```python
# Print numbers from 1 to 10 using a for loop. 
for i in range(1,11):
    print(i) 
```

```python
# Print numbers from 10 down to 1 using a while loop.
num = 10
while num > 0:
    print(num)
    num = num-1
```

```python
# Using a for loop and range(), print all even numbers from 1 to 50.
for i in range(1,51):
    if i%2==0:
        print(i) 
```

```python
# Write a Python program to calculate the sum of numbers from 1 to N.
num = int(input("enter the number"))
total = 0
for i in range(1,num +1):
    total +=i
    print("Sum:",total) 
    
 
```

```python
# Print numbers from 1 to 20, but skip every number divisible by 3.
for i in range(1,21):
    if i %3==0:
        continue
    print(i) 
```

```python
# Write a Python program to print numbers from 1 to 20, but stop the loop when the number reaches 12.
for i in range(1,21):
    if i ==12:
        break
    print(i)
```

```python
# Print this pattern using nested for loops:
# *
# **
# ***
# ****
# *****
n = 5
for i in range(0,n):
    for j in range(0,i+1):
        print("*",end="")
    print() 
```

```python
# Write a Python program to check whether a number is prime or not.
num = int(input("enter the number"))
if num<=1:
    print("Not prime")
else:
    for i in range(2,num):
        if num %i==0:
            print("Not prime")
            break
    else:
        print("Prime")      
```

```python
# Write a Python program to count the number of digits in an integer using a while loop.
number = int(input("enter the number:"))
count = 0
while number>0:
    count +=1
    number = number//10
print(count)     # removes the last digit
```

```python
# Write a Python program to reverse an integer using a while loop.
number = int(input("enter the number:"))
reverse = 0
while number>0:
    digit = number%10  # % 10 → extract last digit
    reverse = reverse*10+digit
    number = number//10
print("Reversed number:",reverse) 
```
