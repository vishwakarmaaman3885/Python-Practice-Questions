#### Practice questions on special magic method and decorator

```python
# Basic — __init__()
# Create a Student class using __init__() that takes name and age and stores them as object attributes.
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
s1 = student("Aman",20)
print(s1.name)
print(s1.age) 
```

```python
# __str__()
# Create a Car class with a brand attribute. Use __str__() so that:
# car = Car("Toyota")
# print(car)

# __init__()  → receives and stores data
# __str__()   → controls what print(object) displays
class car:
    def __init__(self,brand):
        self.brand = brand
    
    def __str__(self):
        return self.brand

c1 = car("Toyota")
c1.brand
```

```python
# Basic — __len__()
# Create a Team class that stores a list of players.
# team = Team(["Aman", "Rahul", "Vikas"])
# print(len(team))

class team:
    def __init__(self,players):
        self.players = players
    def __len__(self):
        return len(self.players)
t1 = team(["Aman","Rahul","Vikas"])
print(len(t1)) 
```

```python
# __add__()
# Create a Number class so that:
class number:
    def __init__(self,value):
        self.value = value
    def __add__(self,other):
        return self.value+ other.value

n1 = number(23)
n2 = number(78)
print(n1+n2) 
```

```python
# @property
# Create a Person class with a private-style attribute _age.
# p = Person(20)
# print(p.age)
class person:
    def __init__(self,age):
        self.__age = age
    @property
    def age(self):
        return self.__age

p1 = person(23)
p1.age
```

```python
# @property + setter
# Create an Employee class with salary.
# Requirements:
# Use @property.
# Use a setter.
# Salary cannot be negative.
# If the user tries employee.salary = -5000, display:
class employee:
    def __init__(self,salary):
        self.__salary = salary
    @property
    def salary(self):
        return self.__salary
    @salary.setter
    def salary(self,salary):
        if salary>=0:
            self.__salary = salary
        else:
            print("salary can't be negative")
e1 = employee(5000)
e1.salary = -5000
print(e1.salary) 
```
