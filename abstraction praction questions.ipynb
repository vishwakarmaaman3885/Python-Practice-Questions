#### Practice questions on abstraction

```python
# Create an abstract class Shape with an abstract method area().Create:Circle,Rectangle
# Both should implement area().
import abc
class Shape(abc.ABC):
    
    @abc.abstractmethod  # Shape(abc.ABC) → abstract class, @abc.abstractmethod → abstract method
    def area(self):
        pass
class Circle(Shape):
    def area(self):
        return 3.14 * 5 * 5
class Rectangle(Shape):
    def area(self):
        return 10 * 4
print(Circle().area())
print(Rectangle().area()) 
```

```python
# Create an abstract class Vehicle with an abstract method start().Create:
# Car → prints "Car starts with a key"
# Bike → prints "Bike starts with a self-start button"
import abc 
class vehicle(abc.ABC):  # abc.ABC → abstract class
    @abc.abstractmethod   # @abc.abstractmethod → abstract method
    def start(self):
        pass
class car(vehicle):
    def start(self):
        print("Car starts with a key")
class bike(vehicle):
    def start(self):
        print("Bike starts with a self-start button")
car().start() 
bike().start()   # car(vehicle) and bike(vehicle) → inheritance 
```

```python
# Create an abstract class Payment with an abstract method pay(amount).
# Create:
# CreditCard
# UPI
import abc
class payment(abc.ABC):
    @abc.abstractmethod
    def pay(self,amount):
        pass
class CreditCard(payment):
    def pay(self,amount):
        print("Paid Rs.",amount,"using credit card")
class UPI(payment):
    def pay(self,amount):
        print("Paid Rs.",amount,"using upi")
c1 = CreditCard()
u1 = UPI()

c1.pay(450)
u1.pay(500) 
```
